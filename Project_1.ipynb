```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
value_to_remove = "Maybe"
df = df[df["Growing_Stress"] != value_to_remove]
```

Remove Timestamp for privacy sake 

```python
df.drop(columns="Timestamp")
```

Stacked Barplot 

```python
palette_colors = ["#9A4EAE", "#6693F5"]
```

```python
grouped = df.groupby(["Occupation", "Growing_Stress"]).size().reset_index(name="Count") 
plt.bar(grouped["Occupation"], grouped["Count"]) 
plt.figure(figsize = (10,6))
sns.barplot(x = "Occupation", y = "Count", data = grouped, hue = "Growing_Stress", palette= palette_colors)
plt.ylabel("Count") 
plt.xlabel("Occupation")
plt.title("Occupation vs Growing Stress")
```

```python
grouped2 = df.groupby(["Days_Indoors", "Growing_Stress"]).size().reset_index(name="Count") 
plt.bar(grouped2["Days_Indoors"], grouped2["Count"]) 
plt.figure(figsize = (10,6))
sns.barplot(x = "Days_Indoors", y = "Count", data = grouped2, hue = "Growing_Stress", palette= palette_colors)
plt.ylabel("Count") 
plt.xlabel("Amount of Days Spent Indoors Consecutively")
plt.title("Amount of Days Spent Indoors Consecutively vs Growing Stress")
```
